```python
import requests
import pandas as pd
import time
import matplotlib.pyplot as plt

def call_dynamo(prompt):
    url="https://api.dynamoai-sec-poc.com/v1/moderation/model/6a9685a1c523d57c1a97eb56/chat/1232abc"
    headers ={
      "Authorization" : "Bearer "  +  key,
       "Content-Type" : "application/json"
    }
    
    payload = {
                        'messages': [{'role': 'user', 'content': prompt, "ragContext": ""}],
                        
                    }
    start = time.perf_counter()
    response =requests.post(url, headers=headers, json=payload)
    ms = (time.perf_counter() - start) * 1000
    response_json=response.json()
    try:
        action=response_json['analyses'][0]['finalAction']
    except:
        action=response_json['message']
    return ms
```

```python
import requests

DYNAMO_API_ENDPOINT = "https://api.dynamoai-sec-poc.com" #TODO - use "https://api.dynamo.ai" if using the DynamoAI SaaS
DYNAMO_API_KEY = key #TODO
POLICY_IDS = ["6a85e3d4dc68968d0863d397",
              "6a85e3d4dc68968d0863d399",
              "6aa49c151d6b4057298393c9",
              "6a9e95dac523d57c1a9a4056",
              "6a831a80b93ee620d4c01e54",
              "6a9e95d844afbd38148e3900",
              "6a9e95d5c523d57c1a9a4048",
              "6a85e3d4dc68968d0863d39b"] # TODO - Multiple policies can be applied at once, but we suggest benchmarking one at a time
MODEL_ID = "<your_model_id>" # OPTIONAL TODO - provide a model id if you want to associate the analysis with a model in the DynamoGuard platform

USER_REQUEST = "What should I invest in?"  # User input


DYNAMO_HEADERS = {
    "Authorization": f"Bearer {DYNAMO_API_KEY}",
    'Content-Type': 'application/json'
}

url = f"{DYNAMO_API_ENDPOINT}/v1/moderation/analyze/"

def call_dynamo_cust(prompt):
    data = {
        "messages": [
            {
                "role": "user", 
                "content": prompt
            },
        ],
        "textType": "MODEL_INPUT",
        "policyIds": POLICY_IDS,
        # "modelId": MODEL_ID -- only needed if you want to associate the logs with a model in the platform
    }

    response = requests.post(url, headers=DYNAMO_HEADERS, json=data)
    try:
        action=response.json()["finalAction"]
    except:
        action=response.json()
    return action
```
